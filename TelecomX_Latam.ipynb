# **🌕🚀Estraccion**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
api = "https://raw.githubusercontent.com/ingridcristh/challenge2-data-science-LATAM/refs/heads/main/TelecomX_Data.json"
datos = pd.read_json(api)
datos.head(10)
```

# **🏭Tranformacion**

```python
#Normalizamos la tabla
df_normalize = pd.json_normalize(datos.to_dict(orient='records'), sep='_')
df_normalize.head()
```

```python
df_normalize.info()
```

```python
#verificamos los datos unico de cada columna
for columna in df_normalize.columns:
    print(f"--- Columna: {columna} ---")
    print(df_normalize[columna].unique())
    print("\n")
```

```python
df_normalize = df_normalize[df_normalize['Churn'] != '']
```

```python
#Comprobamos los clientes
len(df_normalize['customerID'].unique())
```

```python
#convertir el tipo de datos de la columna total en float
df_normalize['account_Charges_Total'] = pd.to_numeric(df_normalize['account_Charges_Total'], errors='coerce')
df_normalize['account_Charges_Total'] = df_normalize['account_Charges_Total'].fillna(0)
```

```python
#Eliminamos la segunda palabra de los nombre de las columnas
columnas_arrgladas = df_normalize.copy()
columnas_arrgladas.columns = columnas_arrgladas.columns.str.split('_').str[-1]
```

```python
columnas_arrgladas.info()
```

```python
#agregamos una nueva columna de cuenta diaria
columnas_arrgladas['Cuentas_Diarias'] = (columnas_arrgladas['Monthly'] / 30).round(2)
```

```python
#traduciendo nombres al español
nombres_espanol = {
    'customerID': 'ID_Cliente',
    'gender': 'Genero',
    'SeniorCitizen': 'Jubilado',
    'Partner': 'Tiene_Pareja',
    'Dependents': 'Tiene_Dependientes',
    'tenure': 'Meses_Permanencia',
    'PhoneService': 'Servicio_Telefonico',
    'MultipleLines': 'Multiples_Lineas',
    'InternetService': 'Tipo_Internet',
    'OnlineSecurity': 'Seguridad_Online',
    'OnlineBackup': 'Copia_Seguridad',
    'DeviceProtection': 'Proteccion_Dispositivo',
    'TechSupport': 'Soporte_Tecnico',
    'StreamingTV': 'Streaming_TV',
    'StreamingMovies': 'Streaming_Peliculas',
    'Contract': 'Tipo_Contrato',
    'PaperlessBilling': 'Facturacion_Digital',
    'PaymentMethod': 'Metodo_Pago',
    'Monthly': 'Pago_Mensual',
    'Total': 'Pago_Total',
    'Churn': 'Abandono'
}
mapa_internet = {'DSL': 'DSL','Fiber optic': 'Fibra óptica','No': 'No tiene'}

mapa_contrato = {'Month-to-month': 'Mes a mes','One year': 'Un año','Two year': 'Dos años'}

mapa_pago = {'Electronic check': 'Cheque electrónico','Mailed check': 'Cheque por correo','Bank transfer (automatic)': 'Transferencia bancaria','Credit card (automatic)': 'Tarjeta de crédito'}

columnas_arrgladas.rename(columns=nombres_espanol, inplace=True)

columnas_arrgladas['Tipo_Internet'] = columnas_arrgladas['Tipo_Internet'].replace(mapa_internet)
columnas_arrgladas['Tipo_Contrato'] = columnas_arrgladas['Tipo_Contrato'].replace(mapa_contrato)
columnas_arrgladas['Metodo_Pago']   = columnas_arrgladas['Metodo_Pago'].replace(mapa_pago)
```

```python
#convirtiendo los valores yes and No en bool
booleanos = {'Yes': 1, 'No': 0}
colums = ['Abandono', 'Tiene_Pareja', 'Tiene_Dependientes', 'Servicio_Telefonico', 'Multiples_Lineas','Proteccion_Dispositivo', 'Soporte_Tecnico', 'Streaming_TV', 'Streaming_Peliculas','Facturacion_Digital' ]


for col in colums:
    columnas_arrgladas[col] = columnas_arrgladas[col].map(booleanos).fillna(0).astype(int)
```

```python
columnas_arrgladas
```

# **🧮📊Carga y analisis**

```python
 # Seleccionamos las columnas numéricas más importantes

resumen_numerico = columnas_arrgladas[['Meses_Permanencia', 'Pago_Mensual', 'Pago_Total']].describe().round(2)
resumen_numerico
```

```python
#creamos una tabla para ver las medias de las columnas mas importantes
tabla_limpia = resumen_numerico.loc[['mean', '50%', 'min', 'max']].T.round(2)
tabla_limpia.columns = ['Promedio', 'Mediana', 'Mínimo', 'Máximo']

nuevos_indices = {
    'Meses_Permanencia': '⏳ Antigüedad (Meses)',
    'Pago_Mensual': '💸 Factura Mensual ($)',
    'Pago_Total': '💰 Total Histórico ($)'
}
tabla_limpia.rename(index=nuevos_indices, inplace=True)

def color_por_fila(row):
    color = 'white'
    if 'Antigüedad' in row.name:
        color = '#e7f5ff'
    elif 'Factura' in row.name:
        color = '#fff5f5'
    elif 'Total' in row.name:
        color = '#ebfbee'
    return [f'background-color: {color}; color: black' for _ in row]


(tabla_limpia.style
    .apply(color_por_fila, axis=1)
    .format('{:,.2f}')
    .set_caption("Resumen: Indicadores Clave")
    .set_properties(**{'border': '1px solid white', 'font-size': '11pt'})
    .set_table_styles([
        {'selector': 'th', 'props': [('background-color', '#333'), ('color', 'white')]}
    ])
)
```

```python
 # Agrupamos por 'Abandono' y calculamos la media de las otras variables
perfil_abandono = columnas_arrgladas.groupby('Abandono')[['Meses_Permanencia', 'Pago_Mensual', 'Pago_Total']].mean().round(2)
perfil_abandono
```

```python
#creamos una tabla para ver una comparativa entre los cliente que se queda y los que abandonan
tabla_abandono = perfil_abandono.copy()

tabla_abandono.index = ['😃 Clientes Leales', '🏃‍♂️ Clientes en Fuga']

(tabla_abandono.style
    .format({'Pago_Mensual': '${:.2f}', 'Pago_Total': '${:.2f}', 'Meses_Permanencia': '{:.1f} meses'})
    .bar(subset=['Meses_Permanencia'], color='#5fba7d', vmin=0)
    .bar(subset=['Pago_Mensual'], color='#d65f5f', vmin=0)
    .set_caption("🚨 Comparativa: ¿Quiénes se van y quiénes se quedan?")
    .set_properties(**{'font-weight': 'bold', 'border-color': 'black'})
)
```

```python
resumen_categorico = columnas_arrgladas[['Tipo_Contrato', 'Metodo_Pago', 'Tipo_Internet']].describe()
resumen_categorico
```

# 📊 Informe: Análisis de Evasión de Clientes (Churn) - Telecom X

---

### 🔹 1. Introducción
**Objetivo:**  La empresa enfrenta una alta tasa de cancelaciones y necesita comprender los factores que llevan a la pérdida de clientes.

**Contexto del Problema:**
Actualmente, la empresa enfrenta una **Tasa de Abandono (Churn Rate) del 26.5%**. Esto significa que 1 de cada 4 clientes termina cancelando el servicio, lo cual representa una fuga de ingresos crítica que debe ser mitigada mediante estrategias basadas en datos.

---

### 🔹 2. Limpieza y Tratamiento de Datos
Para garantizar la calidad del análisis, se procesó el dataset original de **7,043 registros** realizando las siguientes acciones clave:

* **Normalización de Tipos de Datos:** Se corrigió la columna `Total Charges`, convirtiendo valores vacíos a numéricos (0) para evitar errores de cálculo.
* **Traducción y Estandarización:** Se renombraron las columnas y los valores al español, tambien se estandarizaron términos técnicos (ej. *tenure* a *Meses de Permanencia*) para facilitar la lectura del negocio.
* **Codificación Binaria:** Se transformaron variables categóricas (Sí/No) a numéricas (1/0) para facilitar el analisi.
* **Validación:** Se confirmó la inexistencia de valores nulos o duplicados que pudieran tener conflitos con los resultados.

---

### 🔹 3. Análisis Exploratorio de Datos (EDA)
Mediante técnicas de visualización (gráficos de barras, histogramas y boxplots), se evaluaron las dimensiones clave:

1.  **Perfil de Contrato:** Se detectó que la evasión se concentra masivamente en clientes con contrato **"Mes a mes"**, mientras que los contratos a largo plazo muestran alta fidelidad.
2.  **Métodos de Pago:** El método **"Cheque Electrónico"** presenta la mayor fricción y tasa de abandono.
3.  **Tecnología:** Los usuarios de **Fibra Óptica** tienen una tasa de cancelación superior a los de DSL, indicando posibles insatisfacciones con el precio o el servicio.
4.  **Factor Tiempo y Dinero:**
    * **Sensibilidad al Precio:** Los clientes que se van pagan en promedio **$74.44**, frente a los **$61.35** de los leales.
    * **Curva de Vida:** El riesgo de fuga es crítico en los **primeros 6 meses**.

---

### 🔹 4. Conclusiones
El análisis revela un patrón claro de comportamiento:

* **El problema es de Valor Percibido:** La fuga ocurre principalmente en el segmento de **alto valor** (facturas altas y Fibra Óptica). Los clientes no se van por buscar precios bajos, sino porque el servicio no justifica el costo alto.
* **La Fricción impulsa la salida:** La combinación de **Contrato Mensual + Cheque Electrónico** crea una barrera de salida nula, facilitando el abandono ante cualquier inconveniente.
* **El "Triángulo de Riesgo":** Hemos perfilado al cliente propenso al abandono como: *Usuario nuevo (<6 meses) con Fibra Óptica y contrato mensual.*

---

### 🔹 5. Recomendaciones Estratégicas
Basado en los datos, se sugieren las siguientes acciones inmediatas:

> 1.  **Programa de Migración:** Incentivar a los clientes de "Mes a mes" a pasar a contratos anuales mediante descuentos del 5-10%.
> 2.  **Automatización de Pagos:** Ofrecer bonos (GB extra) a quienes cambien de "Cheque Electrónico" a "Tarjeta de Crédito Automática" para reducir la fricción.
> 3.  **Onboarding VIP:** Implementar soporte prioritario para clientes nuevos de Fibra Óptica durante los primeros 6 meses para asegurar su retención temprana.
